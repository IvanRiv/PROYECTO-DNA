# **Tutorial: Creación de entornos virtuales en Windows**

Los entornos virtuales de Python permiten crear espacios independientes para cada proyecto. De esta manera, las librerías y versiones utilizadas en un proyecto no interfieren con las de otros.

En este tutorial aprenderás a crear y utilizar un entorno virtual en **Windows** mediante `venv`.

## **1. Verificar Python**

Abre **CMD** o **PowerShell** y comprueba que Python está instalado:

```bash
python --version
```

También puedes utilizar:

```bash
py --version
```

Si aparece, por ejemplo:

```text
Python 3.12.5
```

Python está instalado correctamente.

## **2. Crear la carpeta del proyecto**

Puedes crear una carpeta desde el explorador de archivos o mediante:

```bash
mkdir mi_proyecto
cd mi_proyecto
```

## **3. Crear el entorno virtual**

Dentro de la carpeta del proyecto ejecuta:

```bash
python -m venv .venv
```

Esto creará una carpeta `.venv` que contendrá el entorno virtual:

```text
mi_proyecto/
│
├── .venv/
└── programa.py
```

## **4. Activar el entorno virtual**

En **PowerShell**:

```powershell
.\.venv\Scripts\Activate.ps1
```

En **CMD**:

```cmd
.venv\Scripts\activate
```

Si se activó correctamente, aparecerá `(.venv)` al inicio de la línea de comandos:

```text
(.venv) C:\Users\Usuario\mi_proyecto>
```

## **5. Instalar librerías**

Con el entorno activo, instala las librerías necesarias:

```bash
pip install numpy pandas matplotlib
```

Puedes comprobar las librerías instaladas con:

```bash
pip list
```

## **6. Crear un archivo de requisitos**

Para guardar las dependencias del proyecto:

```bash
pip freeze > requirements.txt
```

Esto genera un archivo `requirements.txt` con las librerías y versiones utilizadas.

Para instalar posteriormente todas las dependencias:

```bash
pip install -r requirements.txt
```

## **7. Desactivar el entorno**

Cuando termines de trabajar:

```bash
deactivate
```

El indicador `(.venv)` desaparecerá.

> **Nota:** Desactivar el entorno no lo elimina. El entorno puede volver a activarse posteriormente.

## **8. Flujo básico de trabajo**

En un proyecto nuevo, normalmente utilizaremos:

```bash
mkdir mi_proyecto
cd mi_proyecto

python -m venv .venv

# PowerShell
.\.venv\Scripts\Activate.ps1

pip install numpy pandas matplotlib

pip freeze > requirements.txt
```

Cuando terminemos:

```bash
deactivate
```

### **¿Por qué utilizar entornos virtuales?**

Los entornos virtuales permiten que cada proyecto tenga sus propias versiones de Python y librerías. Por ejemplo:

```text
Proyecto A
└── .venv
    └── pandas 2.0

Proyecto B
└── .venv
    └── pandas 2.2
```

Esto evita conflictos entre proyectos y facilita la reproducción de los entornos de desarrollo, algo especialmente importante en **Ciencia de Datos, Machine Learning y MLOps**.


# TUTORIAL: ENTORNO VIRTUAL DE PYTHON EN MAC + JUPYTER

import sys
import subprocess
import os

# 1. Crear el entorno virtual
venv_path = "venv"

if not os.path.exists(venv_path):
    subprocess.run([sys.executable, "-m", "venv", venv_path])
    print("✅ Entorno virtual creado")
else:
    print("ℹ️ El entorno virtual ya existe")

# 2. Determinar las rutas según el sistema
python_venv = os.path.join(venv_path, "bin", "python")
pip_venv = os.path.join(venv_path, "bin", "pip")

# 3. Actualizar pip
subprocess.run([python_venv, "-m", "pip", "install", "--upgrade", "pip"])

# 4. Instalar librerías
paquetes = ["numpy", "pandas", "matplotlib", "jupyter", "ipykernel"]

subprocess.run([pip_venv, "install"] + paquetes)

# 5. Registrar el entorno como kernel de Jupyter
subprocess.run([
    python_venv, "-m", "ipykernel", "install",
    "--user",
    "--name=mi_entorno",
    "--display-name=Python (mi_entorno)"
])

# 6. Mostrar información
print("\n" + "="*50)
print("ENTORNO VIRTUAL CONFIGURADO")
print("="*50)
print(f"Python del entorno: {python_venv}")
print(f"Paquetes instalados: {', '.join(paquetes)}")

# 7. Verificar versión de Python
resultado = subprocess.run(
    [python_venv, "--version"],
    capture_output=True,
    text=True
)

print(f"Versión: {resultado.stdout.strip()}")

print("\n✅ Listo.")
print("En Jupyter selecciona:")
print("Kernel → Change Kernel → Python (mi_entorno)")
